## 2. 张量的创建

张量（Tensor）是 PyTorch 中最基本的数据结构，可以看作是多维数组。下面介绍各种创建方式。

### 2.1 从数据直接创建 `torch.tensor()`

`torch.tensor()` 可以从 Python 列表或元组直接创建张量：

```python
# 标量张量 (0维)
scalar = torch.tensor(3.14)
print(f"标量: {scalar}, 形状: {scalar.shape}, 维度数: {scalar.ndim}")

# 一维张量 (向量)
vec = torch.tensor([1, 2, 3, 4, 5])
print(f"\n向量: {vec}, 形状: {vec.shape}")

# 二维张量 (矩阵)
matrix = torch.tensor([[1, 2, 3],
                       [4, 5, 6]])
print(f"\n矩阵:\n{matrix}, 形状: {matrix.shape}")

# 三维张量 (例如: batch of images)
tensor3d = torch.tensor([[[1, 2], [3, 4]],
                        [[5, 6], [7, 8]]])
print(f"\n三维张量:\n{tensor3d}, 形状: {tensor3d.shape}")
```

### 2.2 特殊张量创建

| 方法 | 说明 |
|------|------|
| `torch.zeros()` | 全零张量 |
| `torch.ones()` | 全一张量 |
| `torch.full()` | 填充指定值 |
| `torch.rand()` | [0, 1) 均匀分布随机张量 |
| `torch.randn()` | 标准正态分布随机张量 |
| `torch.arange()` | 等差序列 |
| `torch.linspace()` | 等间距序列 |
| `torch.eye()` | 单位矩阵 |

```python
# 全零张量
zeros = torch.zeros(2, 3)
print(f"全零张量:\n{zeros}")

# 全一张量
ones = torch.ones(2, 3)
print(f"\n全一张量:\n{ones}")

# 填充指定值
full = torch.full((2, 3), fill_value=7)
print(f"\n填充指定值:\n{full}")

# 均匀随机 [0, 1)
rand = torch.rand(2, 3)
print(f"\n均匀随机:\n{rand}")

# 标准正态随机 N(0, 1)
randn = torch.randn(2, 3)
print(f"\n正态随机:\n{randn}")

# 等差序列
arange = torch.arange(0, 10, 2)
print(f"\n等差序列: {arange}")

# 等间距序列
linspace = torch.linspace(0, 1, 5)
print(f"\n等间距序列: {linspace}")

# 单位矩阵
eye = torch.eye(3)
print(f"\n单位矩阵:\n{eye}")
```

### 2.3 根据已有张量创建

使用 `*_like()` 系列方法可以创建与已有张量**形状相同**的新张量：

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)

print(f"原张量 x:\n{x}, dtype: {x.dtype}")
print(f"zeros_like:\n{torch.zeros_like(x)}")
print(f"ones_like:\n{torch.ones_like(x)}")
print(f"rand_like:\n{torch.rand_like(x)}")

# 也可以显式指定 dtype
y = torch.zeros_like(x, dtype=torch.int32)
print(f"\nzeros_like (int32):\n{y}, dtype: {y.dtype}")
```

## 3. 张量运算

PyTorch 张量支持丰富的数学运算，运算规则与 NumPy 高度一致。

### 3.1 算术运算

支持逐元素加（`+`）、减（`-`）、乘（`*`）、除（`/`）运算：

```python
a = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
b = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)

print(f"a =\n{a}")
print(f"b =\n{b}")

# 逐元素运算
print(f"\na + b =\n{a + b}")
print(f"\na - b =\n{a - b}")
print(f"\na * b =\n{a * b}")      # 逐元素相乘（Hadamard积）
print(f"\na / b =\n{a / b}")

# 幂运算、取模等
print(f"\na ** 2 =\n{a ** 2}")
print(f"\na % 3 =\n{a % 3}")

# 对应函数式写法
print(f"\ntorch.add(a, b) =\n{torch.add(a, b)}")
print(f"torch.mul(a, b) =\n{torch.mul(a, b)}")
print(f"torch.div(a, b) =\n{torch.div(a, b)}")
```

### 3.2 矩阵乘法

矩阵乘法是深度学习中最核心的运算之一。PyTorch 提供了多种方式：

- `torch.mm()`: 二维矩阵乘法
- `torch.bmm()`: 批量二维矩阵乘法
- `torch.matmul()` / `@` 运算符: 通用矩阵乘法（支持高维）

对于矩阵 $A \in \mathbb{R}^{m \times k}$ 和 $B \in \mathbb{R}^{k \times n}$，其乘积 $C = AB \in \mathbb{R}^{m \times n}$ 中，每个元素为：

$$C_{ij} = \sum_{l=1}^{k} A_{il} \cdot B_{lj}$$

```python
A = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)  # (2, 2)
B = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)  # (2, 2)

# 方式1: @ 运算符 (推荐)
C1 = A @ B
print(f"A @ B =\n{C1}")

# 方式2: torch.matmul
C2 = torch.matmul(A, B)
print(f"\ntorch.matmul(A, B) =\n{C2}")

# 方式3: torch.mm (仅支持2D)
C3 = torch.mm(A, B)
print(f"\ntorch.mm(A, B) =\n{C3}")

# 验证: 手动计算
# C[0,0] = 1*5 + 2*7 = 19
# C[0,1] = 1*6 + 2*8 = 22
# C[1,0] = 3*5 + 4*7 = 43
# C[1,1] = 3*6 + 4*8 = 50

# 批量矩阵乘法 torch.bmm
batch_A = torch.randn(3, 2, 4)  # (batch=3, m=2, k=4)
batch_B = torch.randn(3, 4, 5)  # (batch=3, k=4, n=5)
batch_C = torch.bmm(batch_A, batch_B)
print(f"\n批量矩阵乘法: {batch_A.shape} @ {batch_B.shape} = {batch_C.shape}")
```

### 3.3 广播机制 (Broadcasting)

广播机制允许不同形状的张量进行逐元素运算，遵循以下规则：

1. 从最末尾的维度开始比较
2. 两个维度相等，或其中一个为 1，或其中一个不存在 → 兼容
3. 广播后，维度为 1 的维度会沿该轴**复制扩展**

例如，形状 `(3, 1)` 和 `(1, 4)` 可以广播为 `(3, 4)`。

```python
x = torch.arange(3).reshape(3, 1).float()  # (3, 1)
y = torch.arange(4).reshape(1, 4).float()  # (1, 4)

print(f"x 形状: {x.shape}\n{x}")
print(f"\ny 形状: {y.shape}\n{y}")

# 广播: (3,1) + (1,4) → (3,4)
z = x + y
print(f"\nx + y 形状: {z.shape}\n{z}")

# 标量与张量运算
scalar = torch.tensor(10.0)
print(f"\n标量 10 + 向量 [1,2,3] = {scalar + torch.tensor([1.0, 2.0, 3.0])}")

# 不兼容的广播会报错
# a = torch.zeros(3, 4)
# b = torch.zeros(3)  # 报错: (3,4) 与 (3,) 不兼容
```

### 3.4 常用数学函数

```python
x = torch.tensor([0.0, 1.0, 2.0, 3.0, 4.0])

print(f"x = {x}")
print(f"sqrt(x)  = {torch.sqrt(x)}")
print(f"exp(x)   = {torch.exp(x)}")
print(f"log(x+1) = {torch.log(x + 1)}")
print(f"abs(-x)  = {torch.abs(-x)}")
print(f"sin(x)   = {torch.sin(x)}")
print(f"round(x+0.3) = {torch.round(x + 0.3)}")
print(f"clamp(x, 1, 3) = {torch.clamp(x, min=1, max=3)}")
```

### 3.5 聚合运算 (Reduction)

```python
x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print(f"x =\n{x}")

print(f"\nx.sum()    = {x.sum()}")           # 所有元素求和
print(f"x.mean()   = {x.mean()}")           # 均值
print(f"x.max()    = {x.max()}")             # 最大值
print(f"x.min()    = {x.min()}")             # 最小值
print(f"x.argmax() = {x.argmax()}")         # 最大值索引
print(f"x.argmin() = {x.argmin()}")         # 最小值索引

# 按维度聚合
print(f"\nx.sum(dim=0)  = {x.sum(dim=0)}")  # 按行求和 (压缩第0维)
print(f"x.sum(dim=1)  = {x.sum(dim=1)}")    # 按列求和 (压缩第1维)
print(f"x.mean(dim=0) = {x.mean(dim=0)}")   # 按行均值
print(f"x.max(dim=1)  = {x.max(dim=1)}")    # 按列最大值 (返回值和索引)
```

## 4. 索引与切片

PyTorch 张量的索引方式与 NumPy 几乎一致：

```python
x = torch.arange(24).reshape(2, 3, 4).float()
print(f"x 形状: {x.shape}")
print(f"x =\n{x}")
```

```python
# 基本索引
print(f"x[0]      =\n{x[0]}")       # 第一个 "batch"
print(f"\nx[0, 1]  =\n{x[0, 1]}")   # 第0个batch的第1行
print(f"\nx[0, 1, 2] = {x[0, 1, 2]}")  # 具体某个元素

# 切片
print(f"\nx[:, 0, :] =\n{x[:, 0, :]}")  # 所有batch的第0行
print(f"\nx[0, :, ::2] =\n{x[0, :, ::2]}")  # 隔一个取

# 使用索引张量 (高级索引)
indices = torch.tensor([0, 2])
print(f"\nx[0, :, indices] =\n{x[0, :, indices]}")  # 取第0列和第2列

# 布尔索引
mask = x > 10
print(f"\n大于10的元素: {x[mask]}")
```

### ⚠️ 视图 vs 副本

- **基本索引**返回的是**视图(view)**：修改会影响原张量
- **高级索引**（使用张量/布尔索引）返回的是**副本(copy)**

```python
x = torch.arange(6).float()

# 视图 - 共享内存
view_slice = x[1:4]
view_slice[0] = 999
print(f"修改 view_slice 后, x = {x}")  # x[1] 也变成了 999

x = torch.arange(6).float()  # 重置

# 副本 - 独立内存
copy_slice = x[[1, 2, 3]].clone()
copy_slice[0] = 999
print(f"修改 copy_slice 后, x = {x}")  # x 不受影响
```

## 5. 形状变换

形状变换是深度学习中最常用的操作之一。

### 5.1 `view()` 与 `reshape()`

两者功能相似，用于改变张量形状：

| 方法 | 特点 |
|------|------|
| `view()` | 要求张量在内存中是**连续的**，速度更快 |
| `reshape()` | 不要求连续，必要时会复制数据 |

可以使用 `-1` 让 PyTorch 自动推断某个维度的大小。

```python
x = torch.arange(12).float()
print(f"原始 x: {x}, 形状: {x.shape}")

# view - 变为 (2, 6)
x1 = x.view(2, 6)
print(f"\nview(2, 6):\n{x1}")

# view - 自动推断维度
x2 = x.view(3, -1)  # -1 自动计算: 12/3 = 4
print(f"\nview(3, -1):\n{x2}, 形状: {x2.shape}")

# reshape - 同样效果
x3 = x.reshape(4, 3)
print(f"\nreshape(4, 3):\n{x3}")

# 展平为一维
flat = x3.reshape(-1)  # 或 x3.flatten()
print(f"\nflatten: {flat}")

# 多维展平 - 只保留batch维度
batch = torch.randn(8, 3, 32, 32)  # 8张 3x32x32 图像
flat_batch = batch.view(batch.size(0), -1)  # (8, 3072)
print(f"\nbatch flatten: {batch.shape} → {flat_batch.shape}")
```

### 5.2 `squeeze()` 与 `unsqueeze()`

- **`squeeze(dim)`**：移除指定维度（该维度大小必须为1），不指定则移除所有大小为1的维度
- **`unsqueeze(dim)`**：在指定位置**插入**一个大小为1的维度

```python
x = torch.randn(1, 3, 1, 4)
print(f"原始 x 形状: {x.shape}")

# squeeze - 移除所有大小为1的维度
print(f"x.squeeze() 形状: {x.squeeze().shape}")    # (3, 4)

# squeeze - 移除指定维度
print(f"x.squeeze(0) 形状: {x.squeeze(0).shape}")  # (3, 1, 4)
print(f"x.squeeze(2) 形状: {x.squeeze(2).shape}")  # (1, 3, 4)

# unsqueeze - 添加维度
y = torch.randn(3, 4)
print(f"\n原始 y 形状: {y.shape}")
print(f"y.unsqueeze(0) 形状: {y.unsqueeze(0).shape}")  # (1, 3, 4) - 添加batch维
print(f"y.unsqueeze(1) 形状: {y.unsqueeze(1).shape}")  # (3, 1, 4) - 添加通道维
print(f"y.unsqueeze(-1) 形状: {y.unsqueeze(-1).shape}") # (3, 4, 1) - 末尾添加
```

### 5.3 转置与维度交换

| 方法 | 说明 |
|------|------|
| `t()` | 二维转置 |
| `transpose(dim0, dim1)` | 交换两个维度 |
| `permute(*dims)` | 按指定顺序重排所有维度 |
| `T` | `.t()` 的简写（仅2D） |

```python
x = torch.arange(6).reshape(2, 3).float()
print(f"原始 x:\n{x}, 形状: {x.shape}")

# 转置
print(f"\nx.t():\n{x.t()}, 形状: {x.t().shape}")
print(f"x.transpose(0, 1):\n{x.transpose(0, 1)}")

# permute - 多维重排
img = torch.randn(3, 224, 224)  # (C, H, W)
print(f"\n原图 shape (CHW): {img.shape}")
img_hwc = img.permute(1, 2, 0)   # → (H, W, C)
print(f"permute 后 shape (HWC): {img_hwc.shape}")
```

### 5.4 拼接与堆叠

| 方法 | 说明 |
|------|------|
| `torch.cat()` | 沿已有维度拼接 |
| `torch.stack()` | 沿新维度堆叠（所有张量形状必须相同） |

```python
a = torch.tensor([[1, 2], [3, 4]], dtype=torch.float32)
b = torch.tensor([[5, 6], [7, 8]], dtype=torch.float32)

# cat - 沿维度0拼接 (行拼接)
cat0 = torch.cat([a, b], dim=0)
print(f"cat(dim=0):\n{cat0}, 形状: {cat0.shape}")

# cat - 沿维度1拼接 (列拼接)
cat1 = torch.cat([a, b], dim=1)
print(f"\ncat(dim=1):\n{cat1}, 形状: {cat1.shape}")

# stack - 沿新维度堆叠
stacked = torch.stack([a, b], dim=0)
print(f"\nstack(dim=0) 形状: {stacked.shape}\n{stacked}")

stacked1 = torch.stack([a, b], dim=1)
print(f"stack(dim=1) 形状: {stacked1.shape}\n{stacked1}")
```

## 6. Tensor 与 NumPy 互转

Tensor 和 NumPy 数组之间的转换非常方便，且**共享底层内存**（在CPU上）。

```python
# NumPy → Tensor
np_array = np.array([1, 2, 3, 4])
tensor_from_np = torch.from_numpy(np_array)
print(f"NumPy → Tensor: {tensor_from_np}, dtype: {tensor_from_np.dtype}")

# Tensor → NumPy
np_from_tensor = tensor_from_np.numpy()
print(f"Tensor → NumPy: {np_from_tensor}, dtype: {np_from_tensor.dtype}")

# 共享内存演示
print("\n--- 共享内存验证 ---")
np_array[0] = 999
print(f"修改 NumPy 后, Tensor = {tensor_from_np}")  # Tensor 也变了!

tensor_from_np[1] = 888
print(f"修改 Tensor 后, NumPy = {np_array}")  # NumPy 也变了!
```

```python
# 不共享内存的方式
np_array = np.array([1, 2, 3, 4])
tensor_copy = torch.tensor(np_array)  # 使用 torch.tensor() 会复制数据
np_array[0] = 999
print(f"torch.tensor() 创建: {tensor_copy}")  # 不受影响

# GPU 上的 Tensor 不能直接 .numpy()
# 需要先 .cpu()
# gpu_tensor = tensor_copy.cuda()
# gpu_tensor.numpy()  # 报错!
# gpu_tensor.cpu().numpy()  # 正确
```

## 7. 自动微分 (Autograd)

自动微分是 PyTorch 的核心特性之一，它通过**反向传播**自动计算梯度。

### 7.1 核心概念

1. **`requires_grad=True`**：标记张量需要计算梯度
2. **`backward()`**：触发反向传播，计算梯度
3. **`grad`**：存储计算得到的梯度
4. **`torch.no_grad()`**：上下文管理器，禁用梯度计算
5. **`detach()`**：将张量从计算图中分离

```python
# ========== 基础示例: 计算 y = x^2 在 x=3 处的梯度 ==========
x = torch.tensor(3.0, requires_grad=True)
print(f"x = {x}, requires_grad = {x.requires_grad}")

# 前向计算
y = x ** 2
print(f"\ny = x^2 = {y}")

# 反向传播
y.backward()

# 查看梯度: dy/dx = 2x = 6
print(f"\ndy/dx = {x.grad}")
print(f"理论值: dy/dx = 2 * 3 = 6.0  ✓")
```

```python
# ========== 多变量示例: z = (x+y)^2 ===========
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)

z = (x + y) ** 2
print(f"z = (x+y)^2 = {z}")

z.backward()

# dz/dx = 2(x+y) = 10
# dz/dy = 2(x+y) = 10
print(f"\ndz/dx = {x.grad}  (理论: 2*(2+3) = 10) ✓")
print(f"dz/dy = {y.grad}  (理论: 2*(2+3) = 10) ✓")
```

```python
# ========== 链式法则示例: y = sin(x), z = y^2 ===========
x = torch.tensor(1.0, requires_grad=True)

y = torch.sin(x)     # y = sin(1) ≈ 0.8415
z = y ** 2            # z = sin²(1) ≈ 0.7081

z.backward()

# dz/dx = 2*sin(x)*cos(x) = sin(2x) = sin(2) ≈ 0.9093
print(f"z = sin²({x.item()}) = {z.item():.4f}")
print(f"dz/dx = {x.grad.item():.4f}  (理论: sin(2) ≈ 0.9093) ✓")
```

### 7.2 梯度累加陷阱 ⚠️

PyTorch 中梯度是**累加**的，每次调用 `backward()` 都会在之前梯度的基础上叠加。因此在训练循环中必须**清零梯度**：

```python
x = torch.tensor(2.0, requires_grad=True)

# 第一次 backward
y1 = x ** 2
y1.backward()
print(f"第1次 backward: x.grad = {x.grad}")  # 4.0

# 第二次 backward (梯度累加!)
y2 = x ** 3
y2.backward()
print(f"第2次 backward: x.grad = {x.grad}")  # 4.0 + 12.0 = 16.0 (累加了!)

# 正确做法: 每次清零梯度
x.grad.zero_()
y3 = x ** 2
y3.backward()
print(f"清零后 backward: x.grad = {x.grad}")  # 4.0 (正确)
```

### 7.3 `torch.no_grad()` 与 `detach()`

```python
x = torch.tensor(2.0, requires_grad=True)

# ---------- torch.no_grad() ----------
# 在推理/评估时使用，不追踪计算图，节省内存
with torch.no_grad():
    y = x ** 2
    print(f"在 no_grad 中: y.requires_grad = {y.requires_grad}")
# y.backward()  # 报错! y没有grad_fn

# ---------- detach() ----------
# 返回一个新张量，与原张量共享数据，但脱离计算图
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
y_detached = y.detach()
print(f"\ny.requires_grad = {y.requires_grad}")
print(f"y_detached.requires_grad = {y_detached.requires_grad}")
print(f"y_detached 值 = {y_detached}")  # 值相同

# detach 的典型应用: 获取 loss 的数值用于打印
# loss = criterion(output, target)
# print(f"Loss: {loss.detach().item()}")  # 不追踪梯度
```

## 8. 计算图

PyTorch 使用**动态计算图**，每次前向传播都会构建新的计算图。

### 8.1 计算图的组成

- **节点**：张量（数据节点）或函数（运算节点）
- **边**：数据流向，表示张量之间的依赖关系

每个可微张量都有以下属性：
- `.grad_fn`：指向创建该张量的函数（即计算图的边）
- `.next_functions`：前驱函数

### 8.2 查看计算图

```python
# 构建一个简单计算图: z = 2*x + y
x = torch.tensor(2.0, requires_grad=True)
y = torch.tensor(3.0, requires_grad=True)

a = 2 * x          # MulBackward0
b = a + y          # AddBackward0
z = b.relu()       # ReluBackward0

print("计算图追踪:")
print(f"x.grad_fn = {x.grad_fn}")          # None (叶子节点)
print(f"y.grad_fn = {y.grad_fn}")          # None (叶子节点)
print(f"a (2*x).grad_fn = {a.grad_fn}")    # MulBackward0
print(f"b (a+y).grad_fn = {b.grad_fn}")    # AddBackward0
print(f"z (relu).grad_fn = {z.grad_fn}")  # ReluBackward0

# 判断是否是叶子节点
print(f"\nx.is_leaf = {x.is_leaf}")  # True
print(f"y.is_leaf = {y.is_leaf}")  # True
print(f"z.is_leaf = {z.is_leaf}")  # False

# 反向传播
z.backward()
print(f"\ndz/dx = {x.grad}")  # 2 * d(relu(b))/db = 2 * 1 = 2 (因为 b=7 > 0)
print(f"dz/dy = {y.grad}")  # 1 * d(relu(b))/db = 1 * 1 = 1
```

### 8.3 叶子节点与非叶子节点

- **叶子节点 (Leaf)**：用户直接创建的、`requires_grad=True` 的张量
- **非叶子节点**：通过运算产生的中间结果张量

**默认情况下，只有叶子节点的梯度会被保留。** 如需保留非叶子节点的梯度：

```python
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2  # 非叶子节点

y.backward()
print(f"x.grad = {x.grad}")     # 4.0 (保留)
print(f"y.grad = {y.grad}")     # None (默认不保留!)

# 使用 retain_graph=True 和 retain_grad() 保留非叶子梯度
x = torch.tensor(2.0, requires_grad=True)
y = x ** 2
y.retain_grad()  # 显式请求保留 y 的梯度

y.backward(retain_graph=True)  # retain_graph=True 允许再次调用 backward
print(f"\nx.grad = {x.grad}")   # 4.0
print(f"y.grad = {y.grad}")     # 1.0 (现在保留了!)
```

## 9. 综合练习：Autograd 梯度演示

```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  # 设置中文黑体
plt.rcParams['axes.unicode_minus'] = False    # 负号正常显示

# ========== 练习: 验证梯度 ==========
# y = x³ - 3x² + 2x + 1
# dy/dx = 3x² - 6x + 2

x_vals = torch.linspace(-2, 4, 100, requires_grad=True)
y_vals = x_vals ** 3 - 3 * x_vals ** 2 + 2 * x_vals + 1

# 计算梯度
grads = torch.autograd.grad(outputs=y_vals.sum(), inputs=x_vals)[0]

# 绘图
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 函数图像
axes[0].plot(x_vals.detach().numpy(), y_vals.detach().numpy(), 'b-', linewidth=2)
axes[0].set_title(r'$y = x^3 - 3x^2 + 2x + 1$', fontsize=14)
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('y', fontsize=12)
axes[0].grid(True, alpha=0.3)

# 梯度图像
axes[1].plot(x_vals.detach().numpy(), grads.detach().numpy(), 'r-', linewidth=2, label='Autograd 梯度')

# 理论梯度
theoretical_grads = 3 * x_vals ** 2 - 6 * x_vals + 2
axes[1].plot(x_vals.detach().numpy(), theoretical_grads.detach().numpy(), 'g--', linewidth=2, label='理论值: $3x^2-6x+2$')

axes[1].set_title('梯度对比', fontsize=14)
axes[1].set_xlabel('x', fontsize=12)
axes[1].set_ylabel('dy/dx', fontsize=12)
axes[1].legend(fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ========== 模拟梯度下降 ==========
# 目标: 最小化 f(x) = (x-3)²
# 解析解: x* = 3

x = torch.tensor(0.0, requires_grad=True)
lr = 0.1
history_x = [x.item()]
history_f = [(x.item() - 3) ** 2]

for i in range(30):
    f = (x - 3) ** 2
    f.backward()
    x.data -= lr * x.grad  # 手动更新
    x.grad.zero_()          # 清零梯度
    history_x.append(x.item())
    history_f.append((x.item() - 3) ** 2)

# 绘制梯度下降过程
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左图: 函数与下降轨迹
x_plot = torch.linspace(-1, 6, 100)
y_plot = (x_plot - 3) ** 2
axes[0].plot(x_plot, y_plot, 'b-', linewidth=2)
axes[0].plot(history_x, history_f, 'ro-', markersize=5, alpha=0.7, label='梯度下降路径')
axes[0].axvline(x=3, color='g', linestyle='--', alpha=0.5, label='最优解 x=3')
axes[0].set_title('梯度下降过程: $f(x) = (x-3)^2$', fontsize=14)
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('f(x)', fontsize=12)
axes[0].legend(fontsize=11)
axes[0].grid(True, alpha=0.3)

# 右图: 收敛曲线
axes[1].plot(history_f, 'r-', linewidth=2)
axes[1].set_title('损失函数收敛', fontsize=14)
axes[1].set_xlabel('迭代次数', fontsize=12)
axes[1].set_ylabel('f(x)', fontsize=12)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"初始值: x=0, f(x)={history_f[0]:.4f}")
print(f"最终值: x={history_x[-1]:.4f}, f(x)={history_f[-1]:.6f}")
print(f"最优解: x=3.0, f(x)=0.0")
```
