## 一、排序算法

### 1.1 冒泡排序（Bubble Sort）

**原理**：相邻元素两两比较，如果前一个比后一个大就交换，每轮将最大值"冒泡"到末尾。

**时间复杂度**：O(n^2)

````python
def bubble_sort(arr):
n    """冒泡排序"""
n    n = len(arr)
n    # 外层循环：每轮确定一个最大值
n    for i in range(n - 1):
n        swapped = False
n        # 内层循环：相邻比较
n        for j in range(n - 1 - i):
n            if arr[j] > arr[j + 1]:
n                arr[j], arr[j + 1] = arr[j + 1], arr[j]
n                swapped = True
n        # 如果没有交换，说明已经有序
n        if not swapped:
n            break
n    return arr

n# 测试
ndata = [64, 34, 25, 12, 22, 11, 90]
nprint(f"原始数组: {data}")
nresult = bubble_sort(data.copy())
nprint(f"排序结果: {result}")
````

### 1.2 选择排序（Selection Sort）

**原理**：每轮从未排序部分选出最小值，放到已排序部分的末尾。

**时间复杂度**：O(n^2)

````python
def selection_sort(arr):
n    """选择排序"""
n    n = len(arr)
n    for i in range(n - 1):
n        # 找到未排序部分的最小值索引
n        min_idx = i
n        for j in range(i + 1, n):
n            if arr[j] < arr[min_idx]:
n                min_idx = j
n        # 交换到正确位置
n        arr[i], arr[min_idx] = arr[min_idx], arr[i]
n    return arr

ndata = [64, 25, 12, 22, 11]
nprint(f"原始数组: {data}")
nprint(f"排序结果: {selection_sort(data.copy())}")
````

### 1.3 插入排序（Insertion Sort）

**原理**：将元素逐个插入到已排序序列的正确位置（类似打扑克牌时的整理）。

**时间复杂度**：O(n^2)，但对近乎有序的数据效率很高。

````python
def insertion_sort(arr):
n    """插入排序"""
n    n = len(arr)
n    for i in range(1, n):
n        key = arr[i]
n        j = i - 1
n        # 将比key大的元素后移
n        while j >= 0 and arr[j] > key:
n            arr[j + 1] = arr[j]
n            j -= 1
n        arr[j + 1] = key
n    return arr

ndata = [64, 34, 25, 12, 22, 11, 90]
nprint(f"原始数组: {data}")
nprint(f"排序结果: {insertion_sort(data.copy())}")
````

## 二、查找算法

### 2.1 线性查找（Linear Search）

**原理**：从头到尾逐个比较，找到目标值。

**时间复杂度**：O(n)

````python
def linear_search(arr, target):
n    """线性查找：返回目标值的索引，未找到返回-1"""
n    for i, val in enumerate(arr):
n        if val == target:
n            return i
n    return -1

ndata = [11, 12, 22, 25, 34, 64, 90]
nprint(f"数组: {data}")
n
nfor target in [25, 50, 90]:
n    idx = linear_search(data, target)
n    if idx != -1:
n        print(f"找到 {target}，索引为 {idx}")
n    else:
n        print(f"未找到 {target}")
````

### 2.2 二分查找（Binary Search）

**原理**：在**有序数组**中，每次与中间元素比较，缩小搜索范围一半。

**前提条件**：数组必须有序！
**时间复杂度**：O(log n)

````python
def binary_search(arr, target):
n    """二分查找：返回目标值的索引，未找到返回-1"""
n    left, right = 0, len(arr) - 1
n    steps = 0
    
n    while left <= right:
n        steps += 1
n        mid = (left + right) // 2
n        print(f"  第{steps}步: left={left}, mid={mid}, right={right}, arr[mid]={arr[mid]}")
n        
n        if arr[mid] == target:
n            return mid
n        elif arr[mid] < target:
n            left = mid + 1
n        else:
n            right = mid - 1
    
n    return -1

ndata = [11, 12, 22, 25, 34, 64, 90]
nprint(f"有序数组: {data}")
nprint(f"查找 25:")
nresult = binary_search(data, 25)
nprint(f"结果: 索引 {result}\n")

nprint(f"查找 50:")
nresult = binary_search(data, 50)
nprint(f"结果: 未找到")
````

````python
# 练习2 参考答案：递归版二分查找
ndef binary_search_recursive(arr, target, left=0, right=None):
n    if right is None:
n        right = len(arr) - 1
    
n    if left > right:
n        return -1
    
n    mid = (left + right) // 2
n    if arr[mid] == target:
n        return mid
n    elif arr[mid] < target:
n        return binary_search_recursive(arr, target, mid + 1, right)
n    else:
n        return binary_search_recursive(arr, target, left, mid - 1)

ndata = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
nprint(f"数组: {data}")
nfor target in [12, 23, 50]:
n    idx = binary_search_recursive(data, target)
n    status = f"索引{idx}" if idx != -1 else "未找到"
n    print(f"查找{target}: {status}")
````
