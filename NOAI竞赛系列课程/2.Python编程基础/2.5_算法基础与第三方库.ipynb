## 一、排序算法

### 1.1 冒泡排序（Bubble Sort）

**原理**：相邻元素两两比较，如果前一个比后一个大就交换，每轮将最大值"冒泡"到末尾。

**时间复杂度**：O(n^2)

````python
def bubble_sort(arr):
    """冒泡排序"""
    n = len(arr)
    # 外层循环：每轮确定一个最大值
    for i in range(n - 1):
        swapped = False
        # 内层循环：相邻比较
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        # 如果没有交换，说明已经有序
        if not swapped:
            break
    return arr

# 测试
data = [64, 34, 25, 12, 22, 11, 90]
print(f"原始数组: {data}")
result = bubble_sort(data.copy())
print(f"排序结果: {result}")
````

### 1.2 选择排序（Selection Sort）

**原理**：每轮从未排序部分选出最小值，放到已排序部分的末尾。

**时间复杂度**：O(n^2)

````python
def selection_sort(arr):
    """选择排序"""
    n = len(arr)
    for i in range(n - 1):
        # 找到未排序部分的最小值索引
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        # 交换到正确位置
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

data = [64, 25, 12, 22, 11]
print(f"原始数组: {data}")
print(f"排序结果: {selection_sort(data.copy())}")
````

### 1.3 插入排序（Insertion Sort）

**原理**：将元素逐个插入到已排序序列的正确位置（类似打扑克牌时的整理）。

**时间复杂度**：O(n^2)，但对近乎有序的数据效率很高。

````python
def insertion_sort(arr):
    """插入排序"""
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        # 将比key大的元素后移
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

data = [64, 34, 25, 12, 22, 11, 90]
print(f"原始数组: {data}")
print(f"排序结果: {insertion_sort(data.copy())}")
````

### 1.4 三种排序算法对比

````python
import time
import random

# 生成测试数据
random.seed(42)
small = [random.randint(1, 1000) for _ in range(100)]
medium = [random.randint(1, 10000) for _ in range(500)]

def benchmark(sort_func, data, name):
    arr = data.copy()
    start = time.time()
    sort_func(arr)
    elapsed = time.time() - start
    print(f"  {name:<12} {len(data):>4}个元素: {elapsed:.6f}秒")

print("排序算法性能对比:")
for data, label in [(small, "小规模"), (medium, "中等规模")]:
    print(f"\n--- {label} ---")
    benchmark(bubble_sort, data, "冒泡排序")
    benchmark(selection_sort, data, "选择排序")
    benchmark(insertion_sort, data, "插入排序")
    benchmark(sorted, data, "Python内置")
````

## 二、查找算法

### 2.1 线性查找（Linear Search）

**原理**：从头到尾逐个比较，找到目标值。

**时间复杂度**：O(n)

````python
def linear_search(arr, target):
    """线性查找：返回目标值的索引，未找到返回-1"""
    for i, val in enumerate(arr):
        if val == target:
            return i
    return -1

data = [11, 12, 22, 25, 34, 64, 90]
print(f"数组: {data}")

for target in [25, 50, 90]:
    idx = linear_search(data, target)
    if idx != -1:
        print(f"找到 {target}，索引为 {idx}")
    else:
        print(f"未找到 {target}")
````

### 2.2 二分查找（Binary Search）

**原理**：在**有序数组**中，每次与中间元素比较，缩小搜索范围一半。

**前提条件**：数组必须有序！
**时间复杂度**：O(log n)

````python
def binary_search(arr, target):
    """二分查找：返回目标值的索引，未找到返回-1"""
    left, right = 0, len(arr) - 1
    steps = 0
    
    while left <= right:
        steps += 1
        mid = (left + right) // 2
        print(f"  第{steps}步: left={left}, mid={mid}, right={right}, arr[mid]={arr[mid]}")
        
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    
    return -1

data = [11, 12, 22, 25, 34, 64, 90]
print(f"有序数组: {data}")
print(f"查找 25:")
result = binary_search(data, 25)
print(f"结果: 索引 {result}\n")

print(f"查找 50:")
result = binary_search(data, 50)
print(f"结果: 未找到")
````

### 2.3 二分查找 vs 线性查找

````python
import time
import random

# 生成大量有序数据
data_size = 100000
data = list(range(data_size))

# 线性查找
start = time.time()
for _ in range(1000):
    linear_search(data, data_size - 1)  # 查找最后一个元素
linear_time = time.time() - start

# 二分查找
start = time.time()
for _ in range(1000):
    binary_search(data, data_size - 1)
binary_time = time.time() - start

print(f"数据量: {data_size}")
print(f"查找次数: 1000次")
print(f"线性查找耗时: {linear_time:.4f}秒")
print(f"二分查找耗时: {binary_time:.4f}秒")
print(f"\n二分查找比线性查找快约 {linear_time/binary_time:.1f} 倍")
````

````python
# 练习1 参考答案：冒泡排序可视化
def bubble_sort_visual(arr):
    n = len(arr)
    arr = arr.copy()
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
        print(f"第{i+1}轮: {arr}")
    return arr

data = [64, 34, 25, 12, 22]
print(f"原始: {data}\n")
bubble_sort_visual(data)
````

````python
# 练习2 参考答案：递归版二分查找
def binary_search_recursive(arr, target, left=0, right=None):
    if right is None:
        right = len(arr) - 1
    
    if left > right:
        return -1
    
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, right)
    else:
        return binary_search_recursive(arr, target, left, mid - 1)

data = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(f"数组: {data}")
for target in [12, 23, 50]:
    idx = binary_search_recursive(data, target)
    status = f"索引{idx}" if idx != -1 else "未找到"
    print(f"查找{target}: {status}")
````

````python
# 练习3 参考答案：算法对比
import random, time

random.seed(42)
data = [random.randint(1, 10000) for _ in range(1000)]

def time_sort(func, data):
    arr = data.copy()
    t = time.time()
    func(arr)
    return time.time() - t

print(f"数据量: {len(data)}")
print(f"{'算法':<15}{'时间':<15}")
print("-" * 30)
print(f"{'冒泡排序':<13}{time_sort(bubble_sort, data):.4f}秒")
print(f"{'选择排序':<13}{time_sort(selection_sort, data):.4f}秒")
print(f"{'插入排序':<13}{time_sort(insertion_sort, data):.4f}秒")
print(f"{'Python内置':<11}{time_sort(sorted, data):.6f}秒")
````
