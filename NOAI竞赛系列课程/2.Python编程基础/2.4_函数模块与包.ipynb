## 一、自定义函数

### 1.1 基本定义

```python
def 函数名(参数1, 参数2, ...):
    """文档字符串（可选）"""
    函数体
    return 返回值
n```

````python
# 定义和调用函数
ndef greet(name):
n    """向用户打招呼的函数"""
n    print(f"你好, {name}! 欢迎学习Python!")

n# 调用函数
ngreet("张三")
ngreet("NOAI")

n# 带返回值的函数
ndef add(a, b):
n    """返回两个数的和"""
n    return a + b

ndef multiply(a, b):
n    """返回两个数的积"""
n    return a * b

nresult = add(3, 5)
nprint(f"\n3 + 5 = {result}")
nprint(f"3 * 5 = {multiply(3, 5)}")

n# 多个返回值
ndef divide(a, b):
n    """返回商和余数"""
n    return a // b, a % b

nquotient, remainder = divide(17, 5)
nprint(f"\n17 / 5 = {quotient} 余 {remainder}")
````

### 1.2 参数类型

````python
# 1. 位置参数（必须按顺序传入）
ndef power(base, exponent):
n    return base ** exponent

nprint(f"2^10 = {power(2, 10)}")

n# 2. 默认参数（调用时可省略）
ndef introduce(name, age=18, city="北京"):
n    print(f"我叫{name}，今年{age}岁，来自{city}")

nintroduce("张三")
nintroduce("李四", 20)
nintroduce("王五", 22, "上海")

n# 3. 关键字参数（按名称传入，顺序无关）
nintroduce(age=25, name="赵六", city="广州")

n# 4. 可变参数 *args（接收任意多个位置参数）
ndef calculate_sum(*args):
n    print(f"参数: {args}, 类型: {type(args)}")
n    return sum(args)

nprint(f"\nsum(1,2,3) = {calculate_sum(1, 2, 3)}")
nprint(f"sum(1,2,3,4,5) = {calculate_sum(1, 2, 3, 4, 5)}")

n# 5. 可变关键字参数 **kwargs（接收任意多个关键字参数）
ndef print_info(**kwargs):
n    for key, value in kwargs.items():
n        print(f"  {key}: {value}")

nprint("\n学生信息:")
nprint_info(name="张三", age=18, score=95)
````

### 1.3 lambda 表达式（匿名函数）

lambda 用于创建简单的单行函数：
```python
lambda 参数: 表达式
n```

````python
# lambda 表达式
n# 普通函数
ndef square(x):
n    return x ** 2

n# 等价的lambda
nsquare_lambda = lambda x: x ** 2

nprint(f"square(5) = {square(5)}")
nprint(f"square_lambda(5) = {square_lambda(5)}")

n# lambda 的常见用法：配合 sorted(), map(), filter()
nstudents = [("张三", 95), ("李四", 87), ("王五", 92)]
n
n# 按分数排序
nsorted_students = sorted(students, key=lambda s: s[1], reverse=True)
nprint(f"\n按分数排序: {sorted_students}")

n# map: 对每个元素应用函数
nnums = [1, 2, 3, 4, 5]
nsquared = list(map(lambda x: x**2, nums))
nprint(f"平方: {squared}")

n# filter: 过滤满足条件的元素
nevens = list(filter(lambda x: x % 2 == 0, nums))
nprint(f"偶数: {evens}")
````

## 二、变量作用域

````python
# 局部变量与全局变量
nx = 100  # 全局变量

ndef test_scope():
n    x = 50  # 局部变量（遮蔽了全局变量）
n    print(f"函数内 x = {x}")

ntest_scope()
nprint(f"函数外 x = {x}")  # 全局变量不受影响

n# global 关键字修改全局变量
ncounter = 0

ndef increment():
n    global counter
n    counter += 1

nincrement()
nincrement()
nincrement()
nprint(f"\ncounter = {counter}")

n# LEGB 规则
nvar = "global"

ndef outer():
n    var = "enclosing"
n    def inner():
n        var = "local"
n        print(f"inner: {var}")
n    inner()
n    print(f"outer: {var}")

nouter()
nprint(f"global: {var}")
````

## 三、标准库模块

Python提供了丰富的标准库，可以直接使用 `import` 导入。

### 3.1 math 模块

````python
import math

nprint(f"圆周率 pi = {math.pi}")
nprint(f"自然常数 e = {math.e}")
nprint(f"\nsqrt(144) = {math.sqrt(144)}")
nprint(f"2^10 = {math.pow(2, 10)}")
nprint(f"ceil(3.14) = {math.ceil(3.14)}")   # 向上取整
nprint(f"floor(3.14) = {math.floor(3.14)}")  # 向下取整
nprint(f"abs(-42) = {math.fabs(-42)}")        # 绝对值

n# 三角函数
nprint(f"\nsin(pi/2) = {math.sin(math.pi/2):.6f}")
nprint(f"cos(0) = {math.cos(0):.6f}")

n# 对数
nprint(f"\nlog2(1024) = {math.log2(1024)}")
nprint(f"log10(1000) = {math.log10(1000):.4f}")
````

### 3.2 random 模块

````python
import random

n# 设置随机种子（便于复现）
nrandom.seed(42)

nprint(f"randint(1, 10) = {random.randint(1, 10)}")  # 随机整数
nprint(f"random() = {random.random():.4f}")           # 0-1随机浮点数
nprint(f"uniform(1, 100) = {random.uniform(1, 100):.2f}")  # 指定范围浮点数

n# 从序列中随机选择
nfruits = ["苹果", "香蕉", "橙子", "葡萄"]
nprint(f"\nchoice = {random.choice(fruits)}")  # 随机选一个
nprint(f"sample = {random.sample(fruits, 2)}")  # 随机选多个（不重复）

n# 洗牌
ncards = list(range(1, 11))
nprint(f"\n洗牌前: {cards}")
nrandom.shuffle(cards)
nprint(f"洗牌后: {cards}")
````

### 3.3 os 与 sys 模块

````python
import os
import sys

n# os 模块：操作系统相关
nprint("=== os 模块 ===")
nprint(f"当前工作目录: {os.getcwd()}")
nprint(f"操作系统: {os.name}")
n
n# sys 模块：系统相关
nprint("\n=== sys 模块 ===")
nprint(f"Python版本: {sys.version}")
nprint(f"平台: {sys.platform}")
nprint(f"模块搜索路径数量: {len(sys.path)}")

n# 常用：from ... import ...
nfrom math import pi, sqrt
nprint(f"\n直接使用: pi = {pi}, sqrt(9) = {sqrt(9)}")

nfrom datetime import datetime
nprint(f"当前时间: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
````

## 五、练习题

### 练习1：自定义函数
编写函数 `is_palindrome(s)`，判断字符串是否是回文（正读反读都一样）。

### 练习2：递推函数
编写函数 `fibonacci(n)`，返回斐波那契数列的前n项。

### 练习3：标准库练习
使用 `random` 模块生成一个包含10个随机整数的列表，然后使用 `math` 模块计算其统计量。

````python
# 练习1 参考答案：回文判断
ndef is_palindrome(s):
n    s = s.lower().replace(" ", "")
n    return s == s[::-1]

ntest_strings = ["racecar", "hello", "A man a plan a canal Panama", "python"]
nfor s in test_strings:
n    result = "是" if is_palindrome(s) else "不是"
n    print(f"'{s}' {result}回文")
````

````python
# 练习2 参考答案：斐波那契数列
ndef fibonacci(n):
n    """返回斐波那契数列前n项"""
n    if n <= 0:
n        return []
n    elif n == 1:
n        return [0]
n    fib = [0, 1]
n    for i in range(2, n):
n        fib.append(fib[-1] + fib[-2])
n    return fib

nprint(f"前10项: {fibonacci(10)}")
nprint(f"前20项: {fibonacci(20)}")
````

````python
# 练习3 参考答案：统计计算
nimport random, math

nrandom.seed(42)
nnums = [random.randint(1, 100) for _ in range(10)]
nprint(f"随机数: {nums}")

nprint(f"\n和: {sum(nums)}")
nprint(f"平均值: {sum(nums)/len(nums):.1f}")
nprint(f"最大值: {max(nums)}")
nprint(f"最小值: {min(nums)}")
nprint(f"平方根和: {sum(math.sqrt(x) for x in nums):.2f}")
````
