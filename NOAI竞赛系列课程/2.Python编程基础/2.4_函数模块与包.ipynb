## 一、自定义函数

### 1.1 基本定义

```python
def 函数名(参数1, 参数2, ...):
    """文档字符串（可选）"""
    函数体
    return 返回值
```

````python
# 定义和调用函数
def greet(name):
    """向用户打招呼的函数"""
    print(f"你好, {name}! 欢迎学习Python!")

# 调用函数
greet("张三")
greet("NOAI")

# 带返回值的函数
def add(a, b):
    """返回两个数的和"""
    return a + b

def multiply(a, b):
    """返回两个数的积"""
    return a * b

result = add(3, 5)
print(f"\n3 + 5 = {result}")
print(f"3 * 5 = {multiply(3, 5)}")

# 多个返回值
def divide(a, b):
    """返回商和余数"""
    return a // b, a % b

quotient, remainder = divide(17, 5)
print(f"\n17 / 5 = {quotient} 余 {remainder}")
````

### 1.2 参数类型

````python
# 1. 位置参数（必须按顺序传入）
def power(base, exponent):
    return base ** exponent

print(f"2^10 = {power(2, 10)}")

# 2. 默认参数（调用时可省略）
def introduce(name, age=18, city="北京"):
    print(f"我叫{name}，今年{age}岁，来自{city}")

introduce("张三")
introduce("李四", 20)
introduce("王五", 22, "上海")

# 3. 关键字参数（按名称传入，顺序无关）
introduce(age=25, name="赵六", city="广州")

# 4. 可变参数 *args（接收任意多个位置参数）
def calculate_sum(*args):
    print(f"参数: {args}, 类型: {type(args)}")
    return sum(args)

print(f"\nsum(1,2,3) = {calculate_sum(1, 2, 3)}")
print(f"sum(1,2,3,4,5) = {calculate_sum(1, 2, 3, 4, 5)}")

# 5. 可变关键字参数 **kwargs（接收任意多个关键字参数）
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

print("\n学生信息:")
print_info(name="张三", age=18, score=95)
````

### 1.3 lambda 表达式（匿名函数）

lambda 用于创建简单的单行函数：
```python
lambda 参数: 表达式
```

````python
# lambda 表达式
# 普通函数
def square(x):
    return x ** 2

# 等价的lambda
square_lambda = lambda x: x ** 2

print(f"square(5) = {square(5)}")
print(f"square_lambda(5) = {square_lambda(5)}")

# lambda 的常见用法：配合 sorted(), map(), filter()
students = [("张三", 95), ("李四", 87), ("王五", 92)]

# 按分数排序
sorted_students = sorted(students, key=lambda s: s[1], reverse=True)
print(f"\n按分数排序: {sorted_students}")

# map: 对每个元素应用函数
nums = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, nums))
print(f"平方: {squared}")

# filter: 过滤满足条件的元素
evens = list(filter(lambda x: x % 2 == 0, nums))
print(f"偶数: {evens}")
````

## 二、变量作用域

````python
# 局部变量与全局变量
x = 100  # 全局变量

def test_scope():
    x = 50  # 局部变量（遮蔽了全局变量）
    print(f"函数内 x = {x}")

test_scope()
print(f"函数外 x = {x}")  # 全局变量不受影响

# global 关键字修改全局变量
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
increment()
print(f"\ncounter = {counter}")

# LEGB 规则
var = "global"

def outer():
    var = "enclosing"
    def inner():
        var = "local"
        print(f"inner: {var}")
    inner()
    print(f"outer: {var}")

outer()
print(f"global: {var}")
````

## 三、标准库模块

Python提供了丰富的标准库，可以直接使用 `import` 导入。

### 3.1 math 模块

````python
import math

print(f"圆周率 pi = {math.pi}")
print(f"自然常数 e = {math.e}")
print(f"\nsqrt(144) = {math.sqrt(144)}")
print(f"2^10 = {math.pow(2, 10)}")
print(f"ceil(3.14) = {math.ceil(3.14)}")   # 向上取整
print(f"floor(3.14) = {math.floor(3.14)}")  # 向下取整
print(f"abs(-42) = {math.fabs(-42)}")        # 绝对值

# 三角函数
print(f"\nsin(pi/2) = {math.sin(math.pi/2):.6f}")
print(f"cos(0) = {math.cos(0):.6f}")

# 对数
print(f"\nlog2(1024) = {math.log2(1024)}")
print(f"log10(1000) = {math.log10(1000):.4f}")
````

### 3.2 random 模块

````python
import random

# 设置随机种子（便于复现）
random.seed(42)

print(f"randint(1, 10) = {random.randint(1, 10)}")  # 随机整数
print(f"random() = {random.random():.4f}")           # 0-1随机浮点数
print(f"uniform(1, 100) = {random.uniform(1, 100):.2f}")  # 指定范围浮点数

# 从序列中随机选择
fruits = ["苹果", "香蕉", "橙子", "葡萄"]
print(f"\nchoice = {random.choice(fruits)}")  # 随机选一个
print(f"sample = {random.sample(fruits, 2)}")  # 随机选多个（不重复）

# 洗牌
cards = list(range(1, 11))
print(f"\n洗牌前: {cards}")
random.shuffle(cards)
print(f"洗牌后: {cards}")
````

### 3.3 os 与 sys 模块

````python
import os
import sys

# os 模块：操作系统相关
print("=== os 模块 ===")
print(f"当前工作目录: {os.getcwd()}")
print(f"操作系统: {os.name}")

# sys 模块：系统相关
print("\n=== sys 模块 ===")
print(f"Python版本: {sys.version}")
print(f"平台: {sys.platform}")
print(f"模块搜索路径数量: {len(sys.path)}")

# 常用：from ... import ...
from math import pi, sqrt
print(f"\n直接使用: pi = {pi}, sqrt(9) = {sqrt(9)}")

from datetime import datetime
print(f"当前时间: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
````

## 五、练习题

### 练习1：自定义函数
编写函数 `is_palindrome(s)`，判断字符串是否是回文（正读反读都一样）。

### 练习2：递推函数
编写函数 `fibonacci(n)`，返回斐波那契数列的前n项。

### 练习3：标准库练习
使用 `random` 模块生成一个包含10个随机整数的列表，然后使用 `math` 模块计算其统计量。

````python
# 练习1 参考答案：回文判断
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

test_strings = ["racecar", "hello", "A man a plan a canal Panama", "python"]
for s in test_strings:
    result = "是" if is_palindrome(s) else "不是"
    print(f"'{s}' {result}回文")
````

````python
# 练习2 参考答案：斐波那契数列
def fibonacci(n):
    """返回斐波那契数列前n项"""
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib

print(f"前10项: {fibonacci(10)}")
print(f"前20项: {fibonacci(20)}")
````

````python
# 练习3 参考答案：统计计算
import random, math

random.seed(42)
nums = [random.randint(1, 100) for _ in range(10)]
print(f"随机数: {nums}")

print(f"\n和: {sum(nums)}")
print(f"平均值: {sum(nums)/len(nums):.1f}")
print(f"最大值: {max(nums)}")
print(f"最小值: {min(nums)}")
print(f"平方根和: {sum(math.sqrt(x) for x in nums):.2f}")
````
