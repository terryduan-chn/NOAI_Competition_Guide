## 一、顺序结构

**顺序结构**是最简单的程序结构，代码按照从上到下的顺序依次执行。

````python
# 顺序结构示例：计算圆的面积
radius = 5
pi = 3.14159
area = pi * radius ** 2
print(f"半径为 {radius} 的圆，面积为 {area:.2f}")
````

## 二、分支结构（if / elif / else）

```python
if 条件1:
    执行语句1
nelif 条件2:
    执行语句2
nelse:
    执行语句3
n```

> **注意**：Python使用**缩进**（通常是4个空格）来表示代码块！

````python
# 分支结构：成绩评级
score = 85

if score >= 90:
n    grade = "优秀"
nelif score >= 80:
n    grade = "良好"
nelif score >= 70:
n    grade = "中等"
nelif score >= 60:
n    grade = "及格"
nelse:
n    grade = "不及格"
nprint(f"成绩: {score}, 等级: {grade}")
````

### 条件表达式（三元运算符）
```python
值1 if 条件 else 值2
n```

````python
# 条件表达式
nage = 20
nstatus = "成年" if age >= 18 else "未成年"
nprint(f"年龄: {age}, 状态: {status}")

na, b = 15, 23
nmaximum = a if a > b else b
nprint(f"max({a}, {b}) = {maximum}")
````

## 三、循环结构

### 3.1 for 循环

````python
# for 循环遍历列表
nfruits = ["苹果", "香蕉", "橙子", "葡萄"]
nfor fruit in fruits:
n    print(f"我喜欢吃{fruit}")

print("\n--- 使用 enumerate 获取索引 ---")
nfor i, fruit in enumerate(fruits):
n    print(f"第{i+1}种水果: {fruit}")
````

#### range() 函数

| 用法 | 说明 |
|------|------|
| `range(n)` | 生成 0 到 n-1 |
| `range(a, b)` | 生成 a 到 b-1 |
| `range(a, b, step)` | 步长为 step |

````python
# range() 函数的用法
nprint("range(5):", list(range(5)))
nprint("range(2, 8):", list(range(2, 8)))
nprint("range(0, 10, 2):", list(range(0, 10, 2)))
nprint("range(10, 0, -1):", list(range(10, 0, -1)))

n# 用range计算1到100的和
ntotal = sum(range(1, 101))
nprint(f"\n1 + 2 + ... + 100 = {total}")
````

### 3.2 while 循环

````python
# while 循环：倒计时
nprint("倒计时发射！")
ncount = 5
nwhile count > 0:
n    print(count)
n    count -= 1
nprint("发射！")
````

### 3.3 循环控制：break 和 continue

| 关键字 | 作用 |
|--------|------|
| `break` | 立即退出整个循环 |
| `continue` | 跳过本次，进入下一次循环 |

````python
# break：找到第一个能被7整除的数
nprint("--- break 示例 ---")
nfor i in range(1, 100):
n    if i % 7 == 0:
n        print(f"第一个能被7整除的数: {i}")
n        break

n# continue：打印奇数
nprint("\n--- continue: 打印1-10中的奇数 ---")
nfor i in range(1, 11):
n    if i % 2 == 0:
n        continue
n    print(i, end=" ")
nprint()
````

## 四、循环嵌套

### 九九乘法表

````python
# 九九乘法表
nprint("九九乘法表:")
nprint("=" * 50)
nfor i in range(1, 10):
n    for j in range(1, i + 1):
n        print(f"{j}x{i}={i*j:2d}", end="  ")
n    print()
````

### 寻找素数

````python
# 寻找素数
ndef is_prime(n):
n    """判断一个数是否是素数"""
n    if n < 2:
n        return False
n    for i in range(2, int(n**0.5) + 1):
n        if n % i == 0:
n            return False
n    return True

nprimes = [n for n in range(1, 101) if is_prime(n)]
nprint(f"1-100之间的素数共有 {len(primes)} 个:")
nprint(primes)

nprint("\n排列如下:")
nfor i, p in enumerate(primes):
n    print(f"{p:4d}", end="")
n    if (i + 1) % 10 == 0:
n        print()
````

## 五、实践案例：猜数字游戏

````python
# 猜数字游戏
nimport random

nprint("=" * 40)
nprint("       猜数字游戏")
nprint("=" * 40)
nprint("系统已生成一个1-100之间的随机数")

nsecret = 42  # 固定值便于演示
nguesses = [50, 25, 37, 43, 41]  # 模拟猜测序列
n
nprint(f"(答案: {secret}, 模拟猜测过程)")
nfor i, guess in enumerate(guesses, 1):
n    print(f"\n第{i}次猜测: {guess}", end="")
n    if guess == secret:
n        print(f" -> 恭喜！你猜对了！用了{i}次")
n        break
n    elif guess < secret:
n        print(" -> 太小了！")
n    else:
n        print(" -> 太大了！")
````

### 练习4：数字求和
求一个整数的各位数字之和，如 12345 → 15。

````python
# 练习1 参考答案
nscores = [95, 83, 72, 65, 45, 100, 90]
nfor s in scores:
n    if s >= 90: grade = "A"
n    elif s >= 80: grade = "B"
n    elif s >= 70: grade = "C"
n    elif s >= 60: grade = "D"
n    else: grade = "F"
n    print(f"成绩 {s:3d} -> 等级 {grade}")
````

````python
# 练习2 参考答案：打印星号图案
nprint("直角三角形:")
nfor i in range(1, 6):
n    print("*" * i)

print("\n金字塔:")
nfor i in range(1, 6):
n    spaces = " " * (5 - i)
n    stars = "*" * (2 * i - 1)
n    print(spaces + stars)
````

````python
# 练习3 参考答案：百钱买百鸡
nprint("百钱买百鸡的所有方案:")
nprint(f"{'公鸡':<8}{'母鸡':<8}{'小鸡':<8}{'总金额':<8}")
nprint("-" * 32)

nsolutions = []
nfor cock in range(0, 21):
n    for hen in range(0, 34):
n        chick = 100 - cock - hen
n        if chick % 3 != 0:
n            continue
n        if 5 * cock + 3 * hen + chick / 3 == 100:
n            solutions.append((cock, hen, chick))
n            print(f"{cock:<8}{hen:<8}{chick:<8}{5*cock + 3*hen + chick//3:<8}")

nprint(f"\n共找到 {len(solutions)} 种方案")
````

````python
# 练习4 参考答案：数字求和
ndef sum_of_digits(n):
n    n = abs(n)
n    total = 0
n    while n > 0:
n        total += n % 10
n        n //= 10
n    return total

nfor num in [12345, 9999, 0, -567, 1000001]:
n    print(f"{num}: 各位之和 = {sum_of_digits(num)}")
````
