## 一、顺序结构

**顺序结构**是最简单的程序结构，代码按照从上到下的顺序依次执行。

````python
# 顺序结构示例：计算圆的面积
radius = 5
pi = 3.14159
area = pi * radius ** 2
print(f"半径为 {radius} 的圆，面积为 {area:.2f}")
````

## 二、分支结构（if / elif / else）

```python
if 条件1:
    执行语句1
elif 条件2:
    执行语句2
else:
    执行语句3
```

> **注意**：Python使用**缩进**（通常是4个空格）来表示代码块！

````python
# 分支结构：成绩评级
score = 85

if score >= 90:
    grade = "优秀"
elif score >= 80:
    grade = "良好"
elif score >= 70:
    grade = "中等"
elif score >= 60:
    grade = "及格"
else:
    grade = "不及格"
print(f"成绩: {score}, 等级: {grade}")
````

### 条件表达式（三元运算符）
```python
值1 if 条件 else 值2
```

````python
# 条件表达式
age = 20
status = "成年" if age >= 18 else "未成年"
print(f"年龄: {age}, 状态: {status}")

a, b = 15, 23
maximum = a if a > b else b
print(f"max({a}, {b}) = {maximum}")
````

## 三、循环结构

### 3.1 for 循环

````python
# for 循环遍历列表
fruits = ["苹果", "香蕉", "橙子", "葡萄"]
for fruit in fruits:
    print(f"我喜欢吃{fruit}")

print("\n--- 使用 enumerate 获取索引 ---")
for i, fruit in enumerate(fruits):
    print(f"第{i+1}种水果: {fruit}")
````

#### range() 函数

| 用法 | 说明 |
|------|------|
| `range(n)` | 生成 0 到 n-1 |
| `range(a, b)` | 生成 a 到 b-1 |
| `range(a, b, step)` | 步长为 step |

````python
# range() 函数的用法
print("range(5):", list(range(5)))
print("range(2, 8):", list(range(2, 8)))
print("range(0, 10, 2):", list(range(0, 10, 2)))
print("range(10, 0, -1):", list(range(10, 0, -1)))

# 用range计算1到100的和
total = sum(range(1, 101))
print(f"\n1 + 2 + ... + 100 = {total}")
````

### 3.2 while 循环

````python
# while 循环：倒计时
print("倒计时发射！")
count = 5
while count > 0:
    print(count)
    count -= 1
print("发射！")
````

### 3.3 循环控制：break 和 continue

| 关键字 | 作用 |
|--------|------|
| `break` | 立即退出整个循环 |
| `continue` | 跳过本次，进入下一次循环 |

````python
# break：找到第一个能被7整除的数
print("--- break 示例 ---")
for i in range(1, 100):
    if i % 7 == 0:
        print(f"第一个能被7整除的数: {i}")
        break

# continue：打印奇数
print("\n--- continue: 打印1-10中的奇数 ---")
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
````

## 四、循环嵌套

### 九九乘法表

````python
# 九九乘法表
print("九九乘法表:")
print("=" * 50)
for i in range(1, 10):
    for j in range(1, i + 1):
        print(f"{j}x{i}={i*j:2d}", end="  ")
    print()
````

### 寻找素数

````python
# 寻找素数
def is_prime(n):
    """判断一个数是否是素数"""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [n for n in range(1, 101) if is_prime(n)]
print(f"1-100之间的素数共有 {len(primes)} 个:")
print(primes)

print("\n排列如下:")
for i, p in enumerate(primes):
    print(f"{p:4d}", end="")
    if (i + 1) % 10 == 0:
        print()
````

## 五、实践案例：猜数字游戏

````python
# 猜数字游戏
import random

print("=" * 40)
print("       猜数字游戏")
print("=" * 40)
print("系统已生成一个1-100之间的随机数")

secret = 42  # 固定值便于演示
guesses = [50, 25, 37, 43, 41]  # 模拟猜测序列

print(f"(答案: {secret}, 模拟猜测过程)")
for i, guess in enumerate(guesses, 1):
    print(f"\n第{i}次猜测: {guess}", end="")
    if guess == secret:
        print(f" -> 恭喜！你猜对了！用了{i}次")
        break
    elif guess < secret:
        print(" -> 太小了！")
    else:
        print(" -> 太大了！")
````

### 练习4：数字求和
求一个整数的各位数字之和，如 12345 → 15。

````python
# 练习1 参考答案
scores = [95, 83, 72, 65, 45, 100, 90]
for s in scores:
    if s >= 90: grade = "A"
    elif s >= 80: grade = "B"
    elif s >= 70: grade = "C"
    elif s >= 60: grade = "D"
    else: grade = "F"
    print(f"成绩 {s:3d} -> 等级 {grade}")
````

````python
# 练习2 参考答案：打印星号图案
print("直角三角形:")
for i in range(1, 6):
    print("*" * i)

print("\n金字塔:")
for i in range(1, 6):
    spaces = " " * (5 - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
````

````python
# 练习3 参考答案：百钱买百鸡
print("百钱买百鸡的所有方案:")
print(f"{'公鸡':<8}{'母鸡':<8}{'小鸡':<8}{'总金额':<8}")
print("-" * 32)

solutions = []
for cock in range(0, 21):
    for hen in range(0, 34):
        chick = 100 - cock - hen
        if chick % 3 != 0:
            continue
        if 5 * cock + 3 * hen + chick / 3 == 100:
            solutions.append((cock, hen, chick))
            print(f"{cock:<8}{hen:<8}{chick:<8}{5*cock + 3*hen + chick//3:<8}")

print(f"\n共找到 {len(solutions)} 种方案")
````

````python
# 练习4 参考答案：数字求和
def sum_of_digits(n):
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total

for num in [12345, 9999, 0, -567, 1000001]:
    print(f"{num}: 各位之和 = {sum_of_digits(num)}")
````
