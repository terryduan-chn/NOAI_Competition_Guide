## 一、变量与数据类型

在Python中，**变量**是用来存储数据的容器。Python是动态类型语言，不需要事先声明变量类型。

### 1.1 变量命名规则
- 只能包含字母、数字和下划线 `_`
- 不能以数字开头
- 不能使用Python关键字（如 `if`, `for`, `while` 等）
- 区分大小写（`Name` 和 `name` 是不同的变量）

### 1.2 四种基本数据类型

| 类型 | 关键字 | 说明 | 示例 |
|------|--------|------|------|
| 整数 | `int` | 没有小数点的数字 | `42`, `-7`, `0` |
| 浮点数 | `float` | 带小数点的数字 | `3.14`, `-0.5`, `1.0` |
| 布尔值 | `bool` | 逻辑值 | `True`, `False` |
| 字符串 | `str` | 文本数据 | `"hello"`, `"Python"` |

````python
x = 42          # 整数 int
pi = 3.14159     # 浮点数 float
is_student = True  # 布尔值 bool
name = "NOAI"    # 字符串 str

print(type(x))         # 查看数据类型
print(type(pi))
print(type(is_student))
print(type(name))
````

### 1.3 类型转换

使用内置函数可以在不同数据类型之间转换：
- `int()` — 转为整数
- `float()` — 转为浮点数
- `str()` — 转为字符串
- `bool()` — 转为布尔值

````python
# 类型转换示例
a = int("123")      # 字符串 -> 整数
b = float("3.14")   # 字符串 -> 浮点数
c = str(42)         # 整数 -> 字符串
d = bool(0)         # 整数 -> 布尔值

print(f"int('123') = {a}, 类型: {type(a)}")
print(f"float('3.14') = {b}, 类型: {type(b)}")
print(f"str(42) = '{c}', 类型: {type(c)}")
print(f"bool(0) = {d}, bool(1) = {bool(1)}")

# 注意：无法转换的字符串会报错
try:
    x = int("hello")
except ValueError as e:
    print(f"转换错误: {e}")
````

## 二、算术运算符

| 运算符 | 说明 | 示例 | 结果 |
|--------|------|------|------|
| `+` | 加法 | `3 + 2` | `5` |
| `-` | 减法 | `3 - 2` | `1` |
| `*` | 乘法 | `3 * 2` | `6` |
| `/` | 除法（浮点） | `7 / 2` | `3.5` |
| `//` | 整除（向下取整） | `7 // 2` | `3` |
| `%` | 取余（模运算） | `7 % 2` | `1` |
| `**` | 幂运算 | `2 ** 3` | `8` |

````python
a, b = 17, 5

print(f"{a} + {b} = {a + b}")    # 加法
print(f"{a} - {b} = {a - b}")    # 减法
print(f"{a} * {b} = {a * b}")    # 乘法
print(f"{a} / {b} = {a / b}")    # 除法（结果为浮点数）
print(f"{a} // {b} = {a // b}")  # 整除
print(f"{a} % {b} = {a % b}")    # 取余
print(f"{a} ** {b} = {a ** b}")  # 幂运算
````

````python
# 整除的特殊情况
print(f"7 // 2 = {7 // 2}")    # 3
print(f"-7 // 2 = {-7 // 2}")   # -4 (向下取整！)
print(f"7 // -2 = {7 // -2}")   # -4
print(f"7.0 // 2 = {7.0 // 2}")  # 3.0 (结果类型与操作数相关)
````

## 三、关系运算符与逻辑运算符

### 3.1 关系（比较）运算符

| 运算符 | 说明 | 示例 |
|--------|------|------|
| `==` | 等于 | `5 == 5` → `True` |
| `!=` | 不等于 | `5 != 3` → `True` |
| `>` | 大于 | `5 > 3` → `True` |
| `<` | 小于 | `5 < 3` → `False` |
| `>=` | 大于等于 | `5 >= 5` → `True` |
| `<=` | 小于等于 | `5 <= 3` → `False` |

````python
x, y = 10, 20

print(f"x == y: {x == y}")
print(f"x != y: {x != y}")
print(f"x > y:  {x > y}")
print(f"x < y:  {x < y}")
print(f"x >= y: {x >= y}")
print(f"x <= y: {x <= y}")

# 关系运算符也可用于字符串（按字典序比较）
print(f"\n'apple' < 'banana': {'apple' < 'banana'}")
print(f"'abc' == 'abc': {'abc' == 'abc'}")
````

### 3.2 逻辑运算符

| 运算符 | 说明 | 示例 |
|--------|------|------|
| `and` | 与（两者都为True） | `True and False` → `False` |
| `or` | 或（至少一个为True） | `True or False` → `True` |
| `not` | 非（取反） | `not True` → `False` |

````python
# 逻辑运算符
age = 20
has_ticket = True

# and: 两者都为True才为True
print(f"age >= 18 and has_ticket: {age >= 18 and has_ticket}")

# or: 至少一个为True就为True
print(f"age < 18 or has_ticket: {age < 18 or has_ticket}")

# not: 取反
print(f"not has_ticket: {not has_ticket}")

# 组合使用：判断一个年份是否是闰年
year = 2024
is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
print(f"\n{year}年是闰年吗? {is_leap}")
````

## 四、输入输出

### 4.1 print() 函数

````python
# print() 基本用法
print("Hello, NOAI!")
print(42)
print(3.14)
print("姓名:", "张三", "年龄:", 18)

# sep参数：自定义分隔符（默认为空格）
print("2024", "01", "01", sep="-")

# end参数：自定义结尾（默认为换行）
print("第一行", end=" ")
print("第二行")
````

### 4.2 input() 函数

`input()` 用于从键盘接收用户输入，**返回值始终是字符串类型**。

````python
# input() 基本用法（模拟输入过程）
name = "Alice"  # 模拟 input("请输入你的名字: ")
print(f"你好, {name}!")

# 注意：input()返回的是字符串，需要手动转换
age_str = "16"  # 模拟 input("请输入你的年龄: ")
age = int(age_str)  # 转换为整数
print(f"明年你将 {age + 1} 岁")
````

### 4.3 f-string 格式化（推荐）

f-string是Python 3.6+引入的字符串格式化方法，语法简洁直观。

````python
# f-string 格式化
name = "NOAI"
score = 95.5
rank = 1

# 基本用法
print(f"选手 {name} 的得分是 {score}")

# 表达式
print(f"明年的年份是 {2024 + 1}")
print(f"得分是否及格: {score >= 60}")

# 格式控制
pi = 3.14159265
print(f"pi 保留2位小数: {pi:.2f}")
print(f"pi 保留4位小数: {pi:.4f}")
print(f"数字补零: {rank:03d}")  # 001
print(f"百分比: {score/100:.1%}")
````

## 五、实践案例

### 5.1 基础运算练习

````python
# 基础运算练习
a = 2024
b = 7

print(f"a = {a}, b = {b}")
ops = [("+", a+b), ("-", a-b), ("*", a*b), ("/", a/b), ("//", a//b), ("%", a%b)]
for op, r in ops:
    print(f"  {a} {op} {b} = {r}")
````

### 5.2 温度转换器（华氏 ↔ 摄氏）

> 转换公式：
> - 摄氏 → 华氏：`F = C * 9/5 + 32`
> - 华氏 → 摄氏：`C = (F - 32) * 5/9`

````python
# 温度转换器
def celsius_to_fahrenheit(c):
    """摄氏度转华氏度"""
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f):
    """华氏度转摄氏度"""
    return (f - 32) * 5 / 9

print("=" * 50)
print("        温度转换器")
print("=" * 50)

# 摄氏转华氏
celsius = 100
fahrenheit = celsius_to_fahrenheit(celsius)
print(f"\n{celsius}C = {fahrenheit:.1f}F")

# 华氏转摄氏
fahrenheit = 98.6
celsius = fahrenheit_to_celsius(fahrenheit)
print(f"{fahrenheit}F = {celsius:.1f}C")

# 常用温度对照表
print(f"\n{'Celsius':<12}{'Fahrenheit':<12}{'描述'}")
print("-" * 40)
temps = [(0, "水的冰点"), (37, "人体体温"), (100, "水的沸点")]
for c, desc in temps:
    print(f"{c:<12}{celsius_to_fahrenheit(c):<12.1f}{desc}")
````

### 5.3 简单计算器

````python
# 简单计算器
print("=" * 40)
print("       简单计算器")
print("=" * 40)

num1 = 15.0
operator = "/"
num2 = 4.0

if operator == "+":
n    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    result = num1 / num2 if num2 != 0 else "错误：除数不能为0！"
else:
    result = "不支持的运算符！"

print(f"\n计算: {num1} {operator} {num2} = {result}")

# 批量计算
print("\n--- 批量计算 ---")
operations = [(10, "+", 3), (10, "-", 3), (10, "*", 3),
               (10, "/", 3), (10, "//", 3), (10, "%", 3), (2, "**", 10)]
for a, op, b in operations:
n    if op == "+": r = a + b
    elif op == "-": r = a - b
    elif op == "*": r = a * b
    elif op == "/": r = a / b
    elif op == "//": r = a // b
    elif op == "%": r = a % b
    elif op == "**": r = a ** b
    print(f"{a} {op} {b} = {r}")
````
